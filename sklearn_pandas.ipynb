```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler
```

```python
data = {
    "Energy Source" : ["Solar", "Wind", "Hydropower", "Geothermal", "Biomass", "Nuclear"],
    "Energy Consumption" : [1200, np.nan, 2900, np.nan, 2500, 3200], 
    "Cost (Million $)" : [200, 400, np.nan, 150, 250, np.nan] 
}

energy_df = pd.DataFrame(data)

print("Original Energy data with missing value: ")
print(energy_df)
```

```python
cleaned_df = energy_df.dropna()
print(cleaned_df)
print(energy_df)
```

```python
energy_df['Energy Consumption'].fillna(energy_df['Energy Consumption'].mean(), inplace=False)
energy_df['Cost (Million $)'].fillna(energy_df['Cost (Million $)'].mean(), inplace=False)
print(energy_df)
```

```python
forward_fill = energy_df.fillna(method='ffill')
print(forward_fill)
```

```python
backward_fill = energy_df.fillna(method='bfill')
print(backward_fill)
```

```python
energy_df['Missing  Consumption'] = energy_df['Energy Consumption'].isna().astype(int)
print(energy_df)
```

```python
scaler = MinMaxScaler()
energy_df[["Energy Consumption","Cost (Million $)"]] = scaler.fit_transform(energy_df[["Energy Consumption","Cost (Million $)"]])
print(energy_df)
```

```python
scaler = StandardScaler()
energy_df[["Energy Consumption","Cost (Million $)"]] = scaler.fit_transform(energy_df[["Energy Consumption","Cost (Million $)"]])
print(energy_df)
```

```python
energy_encoding_df = pd.get_dummies(energy_df , columns=["Energy Source"])
print(energy_encoding_df)
```
