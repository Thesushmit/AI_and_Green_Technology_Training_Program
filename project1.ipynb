```python
import pygame
import sys

# Initialize Pygame
pygame.init()

# Screen dimensions
WIDTH, HEIGHT = 600, 400
screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Bouncing Elastic Ball")

# Colors
WHITE = (255, 255, 255)
BLUE = (0, 0, 255)

# Clock
clock = pygame.time.Clock()

# Ball properties
ball_radius = 20
x = WIDTH // 2
y = ball_radius
velocity_y = 0.3  # Initial vertical velocity
gravity = 0.1   # Acceleration due to gravity
elasticity = 0.9  # Energy retention on bounce

# Ground level
ground = HEIGHT - ball_radius

# Run the game loop
running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    # Clear screen
    screen.fill(WHITE)

    # Update ball position
    velocity_y += gravity
    y += velocity_y

    # Check for collision with ground
    if y >= ground:
        y = ground  # Reset position to the ground level
        velocity_y = -velocity_y * elasticity  # Reverse velocity and reduce it

    # Draw the ball
    pygame.draw.circle(screen, BLUE, (x, int(y)), ball_radius)

    # Update the display
    pygame.display.flip()

    # Control frame rate
    clock.tick(60)

# Quit Pygame
pygame.quit()
sys.exit()
```
