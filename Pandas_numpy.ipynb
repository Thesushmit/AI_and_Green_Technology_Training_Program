```python
city = [
    {"name" : "cityA" , "Carbon_Footprint" : 500},
    {"name" : "cityB" , "Carbon_Footprint" : 350},
    {"name" : "cityC" , "Carbon_Footprint" : 600},
    {"name" : "cityD" , "Carbon_Footprint" : 200},
]
sustainable_threshold = 400
sus_city = list(filter(lambda city : city["Carbon_Footprint"]<sustainable_threshold, city))
print("sustainable City")
for city in sus_city:
    print(city["name"])
```
