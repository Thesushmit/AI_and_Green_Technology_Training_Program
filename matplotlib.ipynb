```python
import matplotlib.pyplot as plt
```

```python
x = []
y = []
for i in range(10):
    if i % 2 == 0:
        x.append(i)
    else:
        y.append(i*2)
plt.plot(x,y , color = 'g', linestyle='--', marker='*')
plt.title("info")
plt.ylabel("Y axis")
plt.xlabel("X axis")
plt.show()
```

```python
x =   ["Solar", "Wind", "Hydropower", "Geothermal", "Biomass", "Nuclear"]
y = [1200, 10, 2900,10, 2500, 3200]
plt.bar(x,y , color='red')
plt.title("info")
plt.ylabel("Y axis")
plt.xlabel("X axis")
plt.show()
```

```python
Energy_Source = ["Solar", "Wind", "Hydropower", "Geothermal", "Biomass", "Nuclear"]
Cost = [200, 400, 100, 150, 250, 102] 
plt.pie(Cost, labels=Energy_Source, autopct='%1.1f%%')
plt.show()
```

```python
Energy_consumption= [1200, 10, 2900,10, 2500, 3200]
Cost = [200, 400, 100, 150, 250, 102] 
plt.scatter(Cost, Energy_consumption)
plt.show()
```
