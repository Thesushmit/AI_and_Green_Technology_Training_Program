```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "solar" :[1200, 1500, 1200],
    "wind" : [3400, 3600, 2800],
    "hydropower" : [2900, 3100, 2400],
    "biomass" : [2500, 2700, 2400]
} 
df = pd.DataFrame(data)
sns.pairplot(df)
plt.show()
```

```python
corr_mat = df.corr()
sns.heatmap(corr_mat , annot=True, cmap='coolwarm', linewidth=0.9)
plt.title('corr')
plt.show
```

```python
sns.boxplot(data=df)
plt.title("distribution of energy by source")
plt.show()
```

```python
e_v = [100,200,300,400,500]
c_e = [10,20,30,40,50]

df_reg = pd.DataFrame({
    'energy_consumption' : e_v,
    'carbon_emission' : c_e
})
sns.regplot(x='energy_consumption', y='carbon_emission', data=df_reg)
plt.title('regression line')
plt.show()
```

```python
month = ['jan', 'feb', 'march', 'april', 'may', 'june']
df_fac = pd.DataFrame({
    'Month' : month*3,
    'Energy' : [1200,1300,1100,1500, 1400, 1600] *3,
    'Region' : ['north', 'south', 'east'] *6
})
g = sns.FacetGrid(df_fac, col="Region", hue='Region', height=7, aspect=1)
g.map(sns.lineplot, 'Month', 'Energy')
plt.subplots_adjust(top=0.70)
g.fig.suptitle("Energy consumption over time by region")
plt.show()
```
